## Upsert

Upsert is the operation of updating a value if exist or insert otherwise

...  other ways to load data to DB

```python
INSERT INTO humanresources.department (departmentid, "name" , groupname, modifieddate )
VALUES (17, 'Data processing & Engineering', 'Research and Development', CURRENT_DATE)
ON CONFLICT(departmentid) -- specify column of conflict
DO UPDATE SET 
	name = EXCLUDED.name,
	groupname = EXCLUDED.groupname;
```
