# Transactions

```python
BEGIN; -- Start transactions
  UPDATE product p 
  SET color = 'BLACK'
  WHERE p.productid = 316
COMMIT;
```

# Savepoints


```python
BEGIN;
  UPDATE product p 
  SET color = 'BLACK'
  WHERE p.productid = 316;

SAVEPOINT updated_316; -- creates a checkpoint to rollback in case it is necessary

UPDATE product p 
  SET color = 'BLUE'
  WHERE p.productid = 316;

ROLLBACK TO updated_316; -- Set data as it was on checkpoint

UPDATE product p 
  SET color = 'BLUE'
  WHERE p.productid = 317;
COMMIT;
```
