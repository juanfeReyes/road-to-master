# Queries

## Filter

```python
-- fetch all data
SELECT *
FROM product p ;

-- aggregate function
SELECT COUNT(*)
FROM product p ;

-- order and paginate
SELECT *
FROM product p
ORDER BY p.productnumber DESC
LIMIT 20
OFFSET 1;


-- grouping by column and calculate aggregate (Average) of column
SELECT 
	p.daystomanufacture AS daysToBuild,
	AVG(p.safetystocklevel) AS averageStockLevel
FROM product p 
GROUP BY p.daystomanufacture
HAVING p.daystomanufacture > 0;

SELECT 
	p.daystomanufacture AS daysToBuild,
	AVG(p.safetystocklevel) AS averageStockLevel
FROM product p 
GROUP BY p.daystomanufacture
HAVING AVG(p.safetystocklevel) > 500;
```

## Joins

```python
-- JOINS

INSERT INTO productreview (productreviewid, productid, reviewername, reviewdate, emailaddress, rating, "comments")
VALUES (nextval('production.productreview_productreviewid_seq'), 999, 'Juan Felipe Reyes', CURRENT_DATE, 'juan@email.com', 2, 'Test comment');

SELECT 
	p.productid AS ProductID,
	pw.productreviewid AS ViewID
FROM product p 
JOIN productreview pw ON p.productid = pw.productid;

SELECT 
	p.productid AS ProductID,
	pw.productreviewid AS ViewID
FROM product p 
LEFT JOIN productreview pw ON p.productid = pw.productid;

SELECT 
	p.productid AS ProductID,
	pw.productreviewid AS ViewID
FROM product p 
RIGHT JOIN productreview pw ON p.productid = pw.productid;

SELECT 
	p.productid AS ProductID,
	pw.productreviewid AS ViewID
FROM product p 
INNER JOIN productreview pw ON p.productid = pw.productid;

SELECT 
	p.productid AS ProductID,
	pw.productreviewid AS ViewID
FROM product p 
FULL JOIN productreview pw ON p.productid = pw.productid;
```
